### Gaussian 1

```python
import numpy as np
from scipy.stats import multivariate_normal

# Given y2 and y3 observations
observations = np.array([[0.6, 0.1], [-0.4, 0.8], [0.2, 0.5], [0.4, -0.1]])

# Parameters for cluster 1
u1 = np.array([1, 1])
Sigma1 = np.array([[2, 0.5], [0.5, 2]])

gauss1 = multivariate_normal.pdf(observations, mean=u1, cov=Sigma1)
print("Gauss1: ", gauss1)
```

### Gaussian 2

```python
import numpy as np
from scipy.stats import multivariate_normal

# Given y2 and y3 observations
observations = np.array([[0.6, 0.1], [-0.4, 0.8], [0.2, 0.5], [0.4, -0.1]])

# Parameters for cluster 2
u2 = np.array([0, 0])
Sigma2 = np.array([[1.5, 1], [1, 1.5]])

gauss2 = multivariate_normal.pdf(observations, mean=u2, cov=Sigma2)
print("Gauss2: ", gauss2)
```

### Pre-posterior 1

```python
# Initial parameters
pi1 = 0.5
p1 = 0.3
Np1= 1-p1
Gauss1 = np.array([0.0665753,  0.05004889, 0.06837452, 0.05904699])

preposti1 = np.array([])
for i in range(0, 4):
    new = (pi1 * p1 * Gauss1[i]) if i in (0,3) else (pi1 * Np1 * Gauss1[i])
    preposti1 = np.append(preposti1, new)

print("pre-posteriori1: ", preposti1)
```

### Pre-posterior 2

```python
# Initial parameters
pi2 = 0.5
p2 = 0.7
Np2 = 1-p2
Gauss2 = np.array([0.11961837, 0.06819058, 0.12958103, 0.12450009])

preposti2 = np.array([])
for i in range(0, 4):
    new = (pi2 * p2 * Gauss2[i]) if i in (0,3) else (pi2 * Np2 * Gauss2[i])
    preposti2 = np.append(preposti2, new)
print("pre-posteriori2: ", preposti2)
```

### Posteriors

```python
preposti1= np.array([0.0099863,  0.01751711, 0.023931086, 0.00885705])
preposti2= np.array([0.04186643 , 0.01022859 ,  0.01943715, 0.04357503])
posteriori1 = preposti1/(preposti1 + preposti2)
posteriori2 = preposti2/(preposti1 + preposti2)

print("posteriori1: ", posteriori1, "\nposteriori2: ", posteriori2)
```

# M-step

```python
import numpy as np
from scipy.stats import multivariate_normal

#Given Y1 observations
observationsY1 = np.array([1, 0, 0, 1])

# Given y2 and y3 observations
observations = np.array([[0.6, 0.1], [-0.4, 0.8], [0.2, 0.5], [0.4, -0.1]])


#pre calculated posteriors
posteriori1 = np.array([0.19258967, 0.63134504 , 0.55181138 , 0.16892425])
posteriori2 = np.array([0.80741033, 0.36865496 , 0.44818862 , 0.83107575])

#sums
N1=np.sum(posteriori1)
N2=np.sum(posteriori2)

print("N1: ", N1)
print("N2: ", N2)


#update pi's
pi1 = N1/(N1+N2)
pi2 = N2/(N1+N2)

print("\npi1: ", pi1)
print("\npi2: ", pi2)


#update u's
u1 = np.zeros(2, dtype=float)
for i in range(4):
    u1 += posteriori1[i] * observations[i]
u1/=N1
    
u2 = np.zeros(2, dtype=float)
for i in range(4):
    u2 += posteriori2[i] * observations[i]
u2/=N2

print("\nu1: ", u1)
print("\nu2: ", u2)


#update sigmas
sigma1 = np.zeros((2, 2), dtype=float)
for i in range(4):
    sigma1+= posteriori1[i] * np.outer((observations[i] - u1), np.transpose((observations[i] - u1)))
sigma1/=N1

sigma2 = np.zeros((2, 2), dtype=float)
for i in range(4):
    sigma2+= posteriori2[i] * np.outer((observations[i] - u2), np.transpose((observations[i] - u2)))
sigma2/=N2

print("\nsigma1: ", sigma1)
print("\nsigma2: ", sigma2)


#update p's
p1=0
for i in range(4):
    p1+=posteriori1[i] * observationsY1[i]
p1/=N1

p2=0
for i in range(4):
    p2 += posteriori2[i] * observationsY1[i]
p2/=N2

print("\np1: ", p1)
print("\np2: ", p2)
```
